# Extração dos Relatórios - ESJs

O objetivo deste código é extrair os relatórios dos textos referentes aos Extratos de Sessões de Julgamento.

Para isso, usaremos os dados que contém os textos dos processos, e faremos uso de expressões regulares para delimitar o início e o fim do relatório.

```python
import os
import pandas as pd
import re
import random
import numpy as np
import time
```

```python
df = pd.read_csv("D:/Backup Tomas/T0mas/Faculdade/IC/Geral/Github CVM/corporate-fraud-in-brazil/data/interim/textos_dos_processos/textos_esjs.csv")
```

```python
print(df)
```

```python
def extrair_relatorio(texto):

    # Padrão para pegar "Relatório ... É o relatório"
    pattern_relatorio = r"(RELATÓRIO|Relatório|Relatorio|RELATORIO)(.*?)(É o relatório|(?:Rio de Janeiro|São Paulo), (?:\d{2}|\d{1}) de (?:janeiro|fevereiro|março|abril|maio|junho|julho|agosto|setembro|outubro|novembro|dezembro) de 20\d{2})" #Padrão
    match_relatorio = re.search(pattern_relatorio, texto, flags=re.DOTALL) #Procura pelo padrão

    relatorio = (match_relatorio.group(1) + match_relatorio.group(2) + match_relatorio.group(3)) if match_relatorio else np.nan #Definimos o relatório como a união das três strings.

    return relatorio #Retornamos o número do processo e o relatório.
```

```python
df['Relatório'] = df['Texto'].apply(extrair_relatorio)

df
```

Fazemos uma análise, abrindo os pdfs da lista e observamos que os principais problemas se devem às situações em que:
1. O início do relatório é grafado como "r e l a t ó r i o";
2. A palavra "relatório" é grafada sem acento agudo no "o";
3. Em um documento em específico, vemos o início do relatório como "r e l a tó r i o";
4. O relatório, ao invés de terminar com a expressão "é o relatório", termina com a data, no estulo "LUGAR, XX de MÊS de 20XX";
5. O relatório está contido em uma xerox. Para isso, vamos ter que usar outro pacote que lê as imagens.

Vamos ajustar o código para esses casos, mudando a função de acordo com esses pontos.

```python
def extrair_relatorio_2(texto):
   
    # Padrão para pegar "Relatório ... É o relatório"
    pattern_relatorio = r"(relatório|relatorio|r e l a t ó r i o|r e l a t o r i o|r e l a tó r i o)(.*?)(é o relatório|(?:Rio de Janeiro|São Paulo), (?:\d{2}|\d{1}) de (?:janeiro|fevereiro|março|abril|maio|junho|julho|agosto|setembro|outubro|novembro|dezembro) de 20\d{2}|(?:Rio de Janeiro|São Paulo),(?:\d{2}|\d{1}) de (?:janeiro|fevereiro|março|abril|maio|junho|julho|agosto|setembro|outubro|novembro|dezembro) de 20\d{2})" #Padrão
    match_relatorio = re.search(pattern_relatorio, texto, flags=re.DOTALL|re.IGNORECASE) #Procura pelo padrão

    relatorio = (match_relatorio.group(1) + match_relatorio.group(2) + match_relatorio.group(3)) if match_relatorio else np.nan #Definimos o relatório como a união das três strings.

    return relatorio #Retornamos o número do processo e o relatório.
```

```python
df['Relatório2'] = df['Texto'].apply(extrair_relatorio_2)

df
```
