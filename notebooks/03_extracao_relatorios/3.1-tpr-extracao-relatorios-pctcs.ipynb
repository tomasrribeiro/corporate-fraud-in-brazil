O objetivo deste código é extrair os relatórios dos textos referentes aos Pareceres do Comitê de Termo de Compromisso.

Para isso, usaremos os dados que contém os textos dos processos, e faremos uso de expressões regulares para delimitar o início e o fim do relatório.

```python
def extrair_relatorio_2(texto):

    # Padrão para pegar "Relatório ... É o relatório"
    pattern_relatorio = r"(FATOS)(.*?)(CONCLUSÃO|DA CONCLUSÃO)" #Padrão
    match_relatorio = re.search(pattern_relatorio, texto, flags=re.DOTALL) #Procura pelo padrão

    relatorio = (match_relatorio.group(1) + match_relatorio.group(2) + match_relatorio.group(3)) if match_relatorio else np.nan #Definimos o relatório como a união das três strings.

    return relatorio #Retornamos o número do processo e o relatório.
```
